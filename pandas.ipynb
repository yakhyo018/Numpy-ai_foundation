```python

import pandas as pd

```

```python
# 1. DataFrame yaratish
mydataset = {
    "cars": ["BMW", "Volvo", "Ford", "Tesla", "Ferrari", "Ford", "BMW"],
    "counts": [3, 7, 2, 8, 1, 9, 3]
}

df = pd.DataFrame(mydataset)
print(df)
```

```python
# 19. Bo'sh qiymatli DataFrame yaratish
data_with_missing = {
    "cars": ["BMW", "Volvo", None, "Tesla", "Ferrari", None, "Ford"],
    "counts": [3, 7, None, 8, 1, 9, None]
}

df_missing = pd.DataFrame(data_with_missing)
print(df_missing)
```

```python
# 20. Bo'sh qiymatlarni aniqlash
df_missing.isnull()
```

```python
# 21. Har bir ustundagi bo'sh qiymatlar soni
df_missing.isnull().sum()
```

```python
# 22. Bo'sh qiymatli qatorlarni olib tashlash
df_missing.dropna(subset=["cars"])

```

```python
# 23. Bo'sh qiymatlarni to'ldirish
df_missing.fillna({
    "cars": "unknown",
    "counts": 0
})
```
